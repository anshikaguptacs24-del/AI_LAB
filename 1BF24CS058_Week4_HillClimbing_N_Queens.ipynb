```python
def run_nqueens_with_validation():
    try:
        # 1. Ask user for the number of queens
        n_input = input("Enter the number of queens (N, e.g., 8): ")
        try:
            n_queens = int(n_input.strip())
            if n_queens <= 0:
                raise ValueError("Number of queens must be greater than 0.")
        except ValueError as ve:
            print(f"Invalid N. Defaulting to N=8. (Error: {ve})")
            n_queens = 8

        # 2. Ask user for the initial queen positions
        state_input = input(f"Enter initial state as {n_queens} comma-separated row indices (e.g., for N=4: 3, 1, 0, 2): ")
        try:
            # Clean input characters and split
            cleaned_input = state_input.replace(" ", ",").replace(";", ",")
            user_start_state = [int(x.strip()) for x in cleaned_input.split(',') if x.strip() != '']
        except Exception as e:
            print(f"Parsing failed. Using randomized initial state instead. (Error: {e})")
            user_start_state = [random.randint(0, n_queens - 1) for _ in range(n_queens)]

        # 3. Validate dimension match
        if len(user_start_state) != n_queens:
            print(f"[Warning] Expected {n_queens} values, but received {len(user_start_state)}.")
            if len(user_start_state) > 0:
                n_queens = len(user_start_state)
                print(f"Adjusting N to {n_queens} to match input length.")
            else:
                print(f"Defaulting to random configuration for N={n_queens}.")
                user_start_state = [random.randint(0, n_queens - 1) for _ in range(n_queens)]

        # 4. Validate value ranges
        for idx, pos in enumerate(user_start_state):
            if pos < 0 or pos >= n_queens:
                print(f"[Error] Position {pos} at index {idx} is out of bounds for {n_queens}x{n_queens} board.")
                new_pos = max(0, min(pos, n_queens - 1))
                print(f"Correcting position {pos} to {new_pos}.")
                user_start_state[idx] = new_pos

        # 5. Check for overlaps (same row, which is a common invalid initial state for permutations)
        if len(user_start_state) != len(set(user_start_state)):
            print("[Notice] Detected multiple queens placed in the same row (horizontal overlap):")
            row_counts = {}
            for r in user_start_state:
                row_counts[r] = row_counts.get(r, 0) + 1
            for row, count in row_counts.items():
                if count > 1:
                    print(f"  - Row {row} has {count} queens placed on it.")
            print("Proceeding with the solver using this initial setup...")

        print(f"\nInitialized N-Queens with N = {n_queens}.")
        print(f"Initial State: {user_start_state}")
        print("Starting Hill Climbing optimization process...\n")

        # Re-import dependencies in case of environment reset
        import random

        # Re-define local helper within scope for safety
        def get_conflicts_local(state):
            n = len(state)
            conflicts = 0
            for i in range(n):
                for j in range(i + 1, n):
                    if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                        conflicts += 1
            return conflicts

        def get_best_neighbor_local(state):
            n = len(state)
            current_conflicts = get_conflicts_local(state)
            neighbors = []
            best_conflicts = current_conflicts
            for col in range(n):
                original_row = state[col]
                for row in range(n):
                    if row == original_row:
                        continue
                    neighbor = list(state)
                    neighbor[col] = row
                    conflicts = get_conflicts_local(neighbor)
                    if conflicts < best_conflicts:
                        best_conflicts = conflicts
                        neighbors = [neighbor]
                    elif conflicts == best_conflicts:
                        neighbors.append(neighbor)
            if neighbors:
                return random.choice(neighbors), best_conflicts
            return state, current_conflicts

        def print_simple_board_local(state):
            n = len(state)
            for row in range(n):
                print(" ".join("Q" if state[col] == row else "." for col in range(n)))

        # Executing the steps
        state = list(user_start_state)
        current_conflicts = get_conflicts_local(state)
        step = 0
        print(f"Step {step}: {state} | Conflicts: {current_conflicts}")
        print_simple_board_local(state)
        print("-" * 20)

        while True:
            neighbor, neighbor_conflicts = get_best_neighbor_local(state)
            if neighbor_conflicts >= current_conflicts:
                if current_conflicts == 0:
                    print(f"\nSuccess! Solved in {step} steps!")
                else:
                    print(f"\nStuck in a Local Minimum at state {state} with {current_conflicts} conflicts.")
                break
            state = neighbor
            current_conflicts = neighbor_conflicts
            step += 1
            print(f"Step {step}: {state} | Conflicts: {current_conflicts}")
            print_simple_board_local(state)
            print("-" * 20)

    except Exception as general_err:
        print(f"Fatal Execution Error: {general_err}")

run_nqueens_with_validation()
```

```python

def cost(pos):
    n = len(pos)
    h = 0

    for i in range(n):
        for j in range(i + 1, n):
            if pos[i] == pos[j] or abs(pos[i] - pos[j]) == abs(i - j):
                h += 1

    return h


def hill_climbing(pos):
    n = len(pos)

    while True:
        h = cost(pos)
        print("Position:", pos, "Cost:", h)

        if h == 0:
            print("Solution found!")
            return pos

        best = pos
        best_cost = h

        for col in range(n):
            for row in range(n):
                if row != pos[col]:
                    neighbor = pos.copy()
                    neighbor[col] = row
                    neighbor_cost = cost(neighbor)

                    if neighbor_cost < best_cost:
                        best = neighbor
                        best_cost = neighbor_cost

        if best_cost >= h:
            print("Local minimum reached. No solution found.")
            return None

        pos = best.copy()


n = int(input("Enter number of queens: "))

pos = []
print("Enter the row position of each queen:")

for i in range(n):
    row = int(input(f"Queen in column {i}: "))
    if row < 0 or row >= n:
        print("Invalid position!")
        quit()
    pos.append(row)

solution = hill_climbing(pos)

if solution is not None:
    print("Final solution:", solution)
```
