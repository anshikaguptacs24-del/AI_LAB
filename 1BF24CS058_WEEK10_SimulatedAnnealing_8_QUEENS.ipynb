```python

import random
import math

# Calculate number of attacking queen pairs
def calculate_cost(board):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Simulated Annealing algorithm
def simulated_annealing():
    # Take initial state from user
    board = list(map(int, input(
        "Enter 8 queen positions (row values 1-8): "
    ).split()))

    if len(board) != 8 or any(x < 1 or x > 8 for x in board):
        print("Enter exactly 8 values between 1 and 8")
        return

    current_cost = calculate_cost(board)

    T = 1000
    cooling_rate = 0.99

    while T > 0.1 and current_cost != 0:

        # Select a random column
        col = random.randint(0, 7)

        # Select a random row
        new_row = random.randint(1, 8)

        next_board = board.copy()
        next_board[col] = new_row

        next_cost = calculate_cost(next_board)

        delta = current_cost - next_cost

        if delta > 0:
            board = next_board
            current_cost = next_cost

        else:
            probability = math.exp(delta / T)

            if random.random() < probability:
                board = next_board
                current_cost = next_cost

        # Decrease temperature
        T *= cooling_rate

    return board, current_cost


# Main program
result = simulated_annealing()

if result:
    board, cost = result

    print("\nFinal Board:")
    for row in range(1, 9):
        for col in range(8):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

    print("\nNumber of conflicts:", cost)

    if cost == 0:
        print("Solution Found!")
    else:
        print("Solution not found. Try again.")
```
