### A* Algorithm for 8-Puzzle using Misplaced Tiles Heuristic

This implementation uses:
1. **State representation**: A tuple of 9 integers (0 representing the blank space).
2. **Heuristic**: Count of misplaced tiles compared to the goal state (excluding the blank tile).
3. **A* Search**: Uses a priority queue (`heapq`) to expand states with the lowest cost $f(n) = g(n) + h(n)$.

```python
import heapq

class PuzzleState:
    def __init__(self, board, g=0, parent=None):
        self.board = tuple(board)  # 1D tuple of length 9
        self.g = g  # Path cost from start to current node
        self.parent = parent
        self.h = self.calculate_heuristic()
        self.f = self.g + self.h

    def calculate_heuristic(self):
        # Goal state: (1, 2, 3, 4, 5, 6, 7, 8, 0)
        goal = (1, 2, 3, 8,0,4,7,6,5)
        misplaced = 0
        for i in range(9):
            if self.board[i] != 0 and self.board[i] != goal[i]:
                misplaced += 1
        return misplaced

    def get_neighbors(self):
        neighbors = []
        zero_idx = self.board.index(0)
        row, col = zero_idx // 3, zero_idx % 3

        # Define possible moves: Up, Down, Left, Right
        moves = []
        if row > 0: moves.append(-3) # Up
        if row < 2: moves.append(3)  # Down
        if col > 0: moves.append(-1) # Left
        if col < 2: moves.append(1)  # Right

        for move in moves:
            new_zero_idx = zero_idx + move
            new_board = list(self.board)
            # Swap
            new_board[zero_idx], new_board[new_zero_idx] = new_board[new_zero_idx], new_board[zero_idx]
            neighbors.append(PuzzleState(new_board, self.g + 1, self))
        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def solve_a_star(start_board):
    start_state = PuzzleState(start_board)
    goal_board = (1, 2, 3, 8,0,4,7,6,5)

    open_set = []
    heapq.heappush(open_set, start_state)
    closed_set = set()

    nodes_expanded = 0

    while open_set:
        current = heapq.heappop(open_set)

        if current.board == goal_board:
            path = []
            while current:
                path.append(current.board)
                current = current.parent
            return path[::-1], nodes_expanded

        closed_set.add(current.board)
        nodes_expanded += 1

        for neighbor in current.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None, nodes_expanded

def print_board(board):
    for i in range(0, 9, 3):
        print(board[i:i+3])
    print()

# Example execution:
# 0 represents the empty space
start_puzzle = [2,8,3,1,6,4,7,0,5]

print("Initial Board:")
print_board(start_puzzle)

path, expanded = solve_a_star(start_puzzle)
if path:
    print(f"Solved in {len(path) - 1} moves. Nodes expanded: {expanded}\n")
    for step, board in enumerate(path):
        print(f"Step {step}:")
        print_board(board)
else:
    print("No solution found.")
```

```python
import heapq

class PuzzleState:
    def __init__(self, board, g=0, parent=None):
        self.board = tuple(board)  # 1D tuple of length 9
        self.g = g  # Path cost from start to current node
        self.parent = parent
        self.h = self.calculate_heuristic()
        self.f = self.g + self.h

    def calculate_heuristic(self):
        # Goal state: (1, 2, 3, 8, 0, 4, 7, 6, 5)
        # Map target tile values to their goal 2D coordinates (row, col)
        goal_coords = {
            1: (0, 0), 2: (0, 1), 3: (0, 2),
            8: (1, 0), 0: (1, 1), 4: (1, 2),
            7: (2, 0), 6: (2, 1), 5: (2, 2)
        }

        distance = 0
        for i in range(9):
            val = self.board[i]
            if val != 0:  # Exclude the blank tile from distance calculation
                current_row, current_col = i // 3, i % 3
                goal_row, goal_col = goal_coords[val]
                distance += abs(current_row - goal_row) + abs(current_col - goal_col)
        return distance

    def get_neighbors(self):
        neighbors = []
        zero_idx = self.board.index(0)
        row, col = zero_idx // 3, zero_idx % 3

        # Define possible moves: Up, Down, Left, Right
        moves = []
        if row > 0: moves.append(-3) # Up
        if row < 2: moves.append(3)  # Down
        if col > 0: moves.append(-1) # Left
        if col < 2: moves.append(1)  # Right

        for move in moves:
            new_zero_idx = zero_idx + move
            new_board = list(self.board)
            # Swap
            new_board[zero_idx], new_board[new_zero_idx] = new_board[new_zero_idx], new_board[zero_idx]
            neighbors.append(PuzzleState(new_board, self.g + 1, self))
        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def solve_a_star(start_board):
    start_state = PuzzleState(start_board)
    goal_board = (1, 2, 3, 8, 0, 4, 7, 6, 5)

    open_set = []
    heapq.heappush(open_set, start_state)
    closed_set = set()

    nodes_expanded = 0

    while open_set:
        current = heapq.heappop(open_set)

        if current.board == goal_board:
            path = []
            while current:
                path.append(current.board)
                current = current.parent
            return path[::-1], nodes_expanded

        closed_set.add(current.board)
        nodes_expanded += 1

        for neighbor in current.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None, nodes_expanded

def print_board(board):
    for i in range(0, 9, 3):
        print(board[i:i+3])
    print()

# Target board configurations
start_puzzle = [2, 8, 3, 1, 6, 4, 7, 0, 5]

print("Initial Board:")
print_board(start_puzzle)

path, expanded = solve_a_star(start_puzzle)
if path:
    print(f"Solved in {len(path) - 1} moves. Nodes expanded: {expanded}\n")
    for step, board in enumerate(path):
        print(f"Step {step}:")
        print_board(board)
else:
    print("No solution found.")
```
