```python
# First Order Logic - Forward Chaining

# Initial Knowledge Base (Facts)
KB = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

# Forward Chaining Rules
def forward_chaining(KB):

    new_facts = True

    while new_facts:
        new_facts = False
        inferred = set()

        # Rule 1: Enemy(x, America) -> Hostile(x)
        for fact in KB:
            if fact[0] == "Enemy" and fact[2] == "America":
                inferred.add(("Hostile", fact[1]))

        # Rule 2: Missile(x) -> Weapon(x)
        for fact in KB:
            if fact[0] == "Missile":
                inferred.add(("Weapon", fact[1]))

        # Rule 3: Missile(x) AND Owns(A, x) -> Sells(Robert, x, A)
        for fact in KB:
            if fact[0] == "Missile":
                x = fact[1]
                if ("Owns", "A", x) in KB:
                    inferred.add(("Sells", "Robert", x, "A"))

        # Rule 4: American(p) AND Weapon(q)
        # AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)
        for fact in KB:
            if fact[0] == "Sells":
                p, q, r = fact[1], fact[2], fact[3]

                if (("American", p) in KB and
                    ("Weapon", q) in KB and
                    ("Hostile", r) in KB):

                    inferred.add(("Criminal", p))

        # Add newly inferred facts
        for fact in inferred:
            if fact not in KB:
                KB.add(fact)
                print("New Fact Derived:", fact)
                new_facts = True

    return KB


# Execute Forward Chaining
print("Initial Knowledge Base:")
for fact in sorted(KB):
    print(fact)

print("\nForward Chaining Process:\n")

KB = forward_chaining(KB)

# Check Goal
print("\nFinal Knowledge Base:")
for fact in sorted(KB):
    print(fact)

print("\nGoal Checking:")
if ("Criminal", "Robert") in KB:
    print("SUCCESS: Robert is Criminal.")
else:
    print("FAILURE: Goal cannot be proved.")
```
